```python
# Calculate True Positives (TP), False Positives (FP), True Negatives (TN), and False Negatives (FN)
TP = df_proba[(df_proba['True_Label'] == 1) & (df_proba['Predicted_Label'] == 1)].shape[0]
FP = df_proba[(df_proba['True_Label'] == -1) & (df_proba['Predicted_Label'] == 1)].shape[0]
TN = df_proba[(df_proba['True_Label'] == -1) & (df_proba['Predicted_Label'] == -1)].shape[0]
FN = df_proba[(df_proba['True_Label'] == 1) & (df_proba['Predicted_Label'] == -1)].shape[0]

print("True Positives (TP):", TP)
print("False Positives (FP):", FP)
print("True Negatives (TN):", TN)
print("False Negatives (FN):", FN)
```
