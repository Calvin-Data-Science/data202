- In this dataframe, each row refers to a recording of a bird. Column `source_url` provides a link to the page where you can hear the bird.
- Before this lab, we loaded the recordings and generated a **spectrogram** for them, which are in the `spectrogram` row.

> "A spectrogram is a visual representation of the spectrum of frequencies in a signal as it varies with time. Its axes include:
- X-axis (Time): Time progression of the audio signal.
- Y-axis (Frequency): Range of frequencies present in the signal (e.g., low pitch to high pitch).
- Color/Intensity: Indicates the amplitude of each frequency at a given time."

You can take these spectrograms as roughly an "image representation" for the sound file. We will use those as features for our classification!

- You can visualize some of them with this code:

```python
import matplotlib.pyplot as plt
import numpy as np
import random

# Select 3 random spectrograms
num_spectrograms_to_plot = 3
random_indices = random.sample(range(len(birdsong_df)), num_spectrograms_to_plot)

for index in random_indices:
  print(index)
  spectrogram = birdsong_df.iloc[index]['spectrogram']
  plt.figure(figsize=(10, 4))
  plt.imshow(spectrogram, aspect='auto', origin='lower')
  plt.title(f"Spectrogram {index}")
  plt.xlabel("Time")
  plt.ylabel("Frequency")
  plt.colorbar(label="Intensity")
  plt.show()
```

**Adjusted Rand Index (ARI)** and **Normalized Mutual Information (NMI)** are metrics used to evaluate the quality of clustering results by comparing them to a ground truth (true labels).

For example, you can calculate both by doing:

```python
from sklearn.metrics import adjusted_rand_score

# True labels and predicted cluster labels
true_labels = [0, 0, 1, 1, 2, 2]
predicted_clusters = [0, 0, 1, 1, 2, 2]

ari = adjusted_rand_score(true_labels, predicted_clusters)
print("Adjusted Rand Index (ARI):", ari)

from sklearn.metrics import normalized_mutual_info_score

nmi = normalized_mutual_info_score(true_labels, predicted_clusters)
print("Normalized Mutual Information (NMI):", nmi)
```
