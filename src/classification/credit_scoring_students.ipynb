```python
import sklearn.datasets
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix
from sklearn.linear_model import LinearRegression, LogisticRegression
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

Dataset source: https://archive.ics.uci.edu/ml/datasets/Statlog+(German+Credit+Data)

```python
credit_data = sklearn.datasets.fetch_openml('credit-g')
```

```python
print(credit_data.DESCR)
```

```python
credit_df = pd.DataFrame(credit_data.data, columns=credit_data.feature_names)
credit_df.head()
```

```python
credit_df['GOOD'] = credit_data.target
```

```python
pd.crosstab(credit_df['checking_status'], credit_data.target, margins=True)
```

```python
X = pd.get_dummies(
    credit_df[['checking_status']],
    columns=['checking_status'],
    drop_first=True)
X.head()
```

```python
credit_df['GOOD'].head()
```

```python
y = (credit_df['GOOD'] == 'good') + 0.
y.head()
```

```python
X.head()
```

```python
model = LinearRegression()
model.fit(X, y)
```

```python
predictions = model.predict(X)
```

```python
plt.hist(predictions)
```

```python
pd.Series(predictions).value_counts()
```

```python
offer_loan = predictions > .5
confusion_matrix(y, offer_loan)
```
